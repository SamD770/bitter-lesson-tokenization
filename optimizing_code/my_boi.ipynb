```python
from transformers import AutoTokenizer

my_tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")

train_batch_sequences = [
    "This is a list",
    "of strings",
    "of varying lengths, with",
    "padding"
]


data_proto = my_tokenizer(train_batch_sequences, padding=True, return_tensors="pt")
data_proto['attention_mask']
```
